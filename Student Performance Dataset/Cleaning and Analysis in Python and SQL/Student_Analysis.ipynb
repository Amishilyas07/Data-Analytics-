# No of Male and Female By Major 

```python
Gender_BY_Major  = pd.read_sql_query('select Major,Gender,Count(StudentID) AS Gender_Count from [student_performance_data] group by Major,Gender',con=connection)
Gender_BY_Major
```

```python
Gender_By_Major_Count = Gender_BY_Major.pivot(index='Major',columns='Gender',values='Gender_Count').reset_index()
```

```python
Gender_By_Major_Count
```

```python
Major = Gender_By_Major_Count['Major']
Female = Gender_By_Major_Count['Female']
Male = Gender_By_Major_Count['Male']
```

```python
x_indexes = np.arange(len(Major))
width=0.2
colors = ['#00b4d8','#ffc8dd']
plt.title('No Students By Major',fontdict={'fontweight':'bold'})
plt.bar(x_indexes-width,Male,width=width*1.5,label='Male',color='#00b4d8')
plt.bar(x_indexes+width,Female,width=width*1.5,label='Female',color='#ffc8dd')
plt.legend()
plt.xticks(ticks=x_indexes,labels=Major)
plt.xlabel('Major',fontdict={'fontweight':'bold'})
plt.ylabel('Count',fontdict={'fontweight':'bold'})
plt.show()
```
