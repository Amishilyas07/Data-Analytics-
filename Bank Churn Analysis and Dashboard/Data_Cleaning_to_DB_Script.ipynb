```python
df1.drop_duplicates(inplace=True)
df2.drop_duplicates(inplace=True)
```

```python
def Age_Grouping(x):
    if 18 <= x <= 25:
        return '18-25'
    elif 26 <= x <= 35:
        return '26-35'
    elif 36 <= x <= 45:
        return '36-45'
    elif 46 <= x <= 55:
        return '46-55' 
    elif 56 <= x <= 65:
        return '56-65'
    elif 66 <= x <= 75:
        return '66-75'
    elif 76 <= x <= 85:
        return '76-85'
    else:
        return '86+'
```

```python
df1['Age_Groups'] = df1['Age'].apply(Age_Grouping)
```

```python
def CreditScore_Status(score):
        if score < 580:
            return "Poor"
        elif 580 <= score < 670:
            return "Fair"
        elif 670 <= score < 740:
            return "Good"
        else:
            return "Excellent"
```

```python
df1['Credit_Score_Status'] = df1['CreditScore'].apply(CreditScore_Status)
```

```python
def categorize_balance(balance):
    if balance == 0:
        return "0"
    elif balance <= 50000:
        return "1-50000"
    elif balance <= 150000:
        return "50001-150000"
    else:
        return "150001+"
```

```python
df2['Balance_Category'] = df2['Balance'].apply(categorize_balance)
```
