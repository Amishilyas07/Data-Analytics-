## **KPI (Key Performance Indicators) for Dashboard**
Since this dataset is likely related to customer retention, churn analysis, and financial services, here are some **KPIs** you can calculate and visualize in your dashboard:

### **1. Customer Retention & Churn**
- **Churn Rate (%):**  
  \[
  \text{Churn Rate} = \left( \frac{\text{Number of Exited Customers}}{\text{Total Customers}} \right) \times 100
  \]
- **Retention Rate (%):**  
  \[
  \text{Retention Rate} = 100 - \text{Churn Rate}
  \]

### **2. Financial Metrics**
- **Average Account Balance:**  
  \[
  \text{Avg Balance} = \frac{\sum \text{Balance}}{\text{Total Customers}}
  \]
- **Average Credit Score:**  
  \[
  \text{Avg Credit Score} = \frac{\sum \text{CreditScore}}{\text{Total Customers}}
  \]
- **Total Deposits:**  
  \[
  \text{Total Deposits} = \sum \text{Balance}
  \]
- **Average Estimated Salary:**  
  \[
  \text{Avg Salary} = \frac{\sum \text{EstimatedSalary}}{\text{Total Customers}}
  \]

### **3. Customer Engagement**
- **Active vs. Inactive Customers:** Count the number of customers who are active (`IsActiveMember = 1`) vs. inactive (`IsActiveMember = 0`).
- **Customers with Credit Cards:**  
  \[
  \text{Percentage of Customers with Credit Cards} = \left( \frac{\text{Customers with Credit Cards}}{\text{Total Customers}} \right) \times 100
  \]
- **Average Tenure:**  
  \[
  \text{Avg Tenure} = \frac{\sum \text{Tenure}}{\text{Total Customers}}
  \]

### **4. Product Analysis**
- **Distribution of Number of Products:** How many customers use 1, 2, 3, or 4 products?
- **Churn Rate by Number of Products:** Which product count has the highest churn?
- **Product Adoption Rate:**  
  \[
  \text{Adoption Rate} = \left( \frac{\text{Customers with More than 1 Product}}{\text{Total Customers}} \right) \times 100
  \]

### **5. Churn Analysis**
- **Churn Rate by Country:** Which country has the highest churn?
- **Churn Rate by Credit Score:** Do lower credit scores correlate with higher churn?
- **Churn Rate by Age Group:** Which age group is churning the most?
- **Churn Rate by Balance:** Do customers with higher balances churn less?
