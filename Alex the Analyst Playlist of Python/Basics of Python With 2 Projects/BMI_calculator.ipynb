```python
# BMI = (weight in pounds x 703) / (height in inches  x height in inches)

weight = float(input("Enter Your Weight in pounds : "))
height = float(input("Enter Your Height in inches : "))

BMI = (weight * 703)/ (height * height)

print(f"Your BMI : {BMI}") 
```

```python
# Correct
name = input("Enter Your Name")
print(f"Your Name  : {name}")
weight = float(input("Enter Your Weight in Pounds : "))
print(f"Your Weight in Pounds : {weight}")
height = float(input("Enter Your Height in inches : "))
print(f"Your Height in inches : {height}")

BMI = (weight * 703)/ (height * height)
print(f"Your BMI : {BMI}")
if(BMI>0):
    if(BMI<18.5):
        print(f"You Are Under Weight {name}")
    elif(BMI<=24.9) :
        print(f"Your Are Normal Weight {name}")
    elif(BMI<=29.9) :
        print(f"Your are Overweight {name}")
    elif(BMI<=34.9) :
        print(f"Your are Obese {name}")
    elif(BMI<=39.9) :
        print(f"Severly Obese {name}")    
    elif(BMI>=40) :
        print(f"Extremely Obese {name}")                   
    else:print("Enter Valid input")
```
