# Functions in Python

```python
def myFun():
    print("Hello This is function")
```

```python
myFun()
```

```python
def myNumbers(a,b):
    print(a,b)
```

```python
myNumbers(5,2)
```

```python
def squareFun(number):
    print(number*number)
```

```python
squareFun(5)
```

```python
def numPowerFun(number,Power):
    print(number**Power)
```

```python
numPowerFun(2,2)
```

```python
numPowerFun(2,8)
```

```python
# arbitary function

def myFunction (**number):
    print("My number is " + number['integer'])
```

```python
myFunction(integer='786')
```

```python
myFunction(integer='222')
```

```python
def myArbitaryFunction(**number):
    print('My number is ' + number['integer'] + 'my other number is ' + number['integer2'])
```

```python
myArbitaryFunction(integer='125',integer2='200')
```

```python
myArbitaryFunction(integer = '250', integer2='555')
```
