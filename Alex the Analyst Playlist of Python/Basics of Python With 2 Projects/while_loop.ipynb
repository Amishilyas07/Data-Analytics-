# While Loop in Python

```python
number = 5

while number >=1:
    print(number)
    number = number-1
```

```python
number = 1
while number <=5:
    print(number)
    number=number+1
```

```python
# break statement
number = 1
while number <=5:
    print(number)
    number=number+1
    if number >3:
        break
```

```python
number = 1
while number <=5:
    print(number)
    if number ==6:
        break
    number = number+1
else:
    print("Program Ended")
```

```python
number = 0 
while number <5:
    number=number+1
    if number ==3:
        continue
    print(number)
```
