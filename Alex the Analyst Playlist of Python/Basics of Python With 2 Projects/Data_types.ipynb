```python
# Sets
my_Set = {1,1,2,5} # Cannot Have Duplication
print(my_Set)
```

```python
myset1 = {1,1,2,4,6,7}
myset2 = {1,2,5,7,4,8,10}
print(myset1 | myset2) # Taking Union
```

```python
print(myset1 & myset2) # Intersection
```

```python
print(myset1 - myset2) # Uncommon in set1
```

```python
print(myset2-myset1) # uncommon in set2
```

```python
print(myset1 ^ myset2) # unique or uncommon values in both sets
```
