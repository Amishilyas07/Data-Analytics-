```python
file_name = os.listdir(path)
```

```python
folder_name = ['csv_files','jpg_files','png_files']

for i in range(0,3):
 if not os.path.exists(path + folder_name[i]):
  print(path + folder_name[i])
  os.makedirs(path +folder_name[i])
```

```python
for file in file_name:
    try:
        if ".csv" in file and not os.path.exists('csv_files/' + file):
         shutil.move(path + file,path + 'csv_files/' + file)
        elif  ".png" in file and not os.path.exists('png_files/' + file):
         shutil.move(path + file,path + 'png_files/' + file)
        elif  ".jpg" in file and not os.path.exists('jpg_files/' + file):
         shutil.move(path + file,path + 'jpg_files/' + file)
    except FileNotFoundError as e:
        print(f"{e}")
```
