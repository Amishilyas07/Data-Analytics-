# For Loop

```python
my_list = [1,6,4,7,22]
for i in my_list:
    print(i)
```

```python
my_list = [1,6,4,7,22]
for i in my_list:
    print(i*2)
```

```python
my_list = [1,6,4,7,22]
for i in my_list:
    print("Number")
```

```python
my_list = [1,6,4,7,22]
for i in my_list:
    print(i+i)
```

```python
my_dictionary = {'name':'Amish','Age':20,'Car':'Bugtti','fav_Games':['FIFA','Tekken']}
for i in my_dictionary:
    print(i)
```

```python
for i in my_dictionary.values():
    print(i)
```

```python
for key,values in my_dictionary.items():
    print(key,values)
```

```python
for key , values in my_dictionary.items():
    print(key ,'->' ,values)
```

```python
# Nested for Loops
my_list1 = ['Name','Age']
my_list2 = ['Amish',22]
for i in my_list1:
    for j in my_list2:
        print(i,' is' , j)
```

```python
my_list1 = [1,2,3,4,5]
my_list2 = [6,7,8,9,10]
for i in my_list1:
    for j in my_list2:
        print(i,j)
```

```python
my_list1 = [1,2,3,4,5]
my_list2 = [6,7,8,9,10]
for i in my_list1:
    for j in my_list2:
        print(i)
```
