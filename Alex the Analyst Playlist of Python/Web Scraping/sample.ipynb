```python
from bs4 import BeautifulSoup
import requests
```

```python
url_of_Website = 'https://en.wikipedia.org/wiki/List_of_highest-grossing_Indian_films'
```

```python
request  = requests.get(url_of_Website)
```

```python
print(request) 
```

```python
soup = BeautifulSoup(request.text,'html.parser')
```

```python
myrows = soup.find('tbody').find_all('tr')[1:]
```

```python
print(myrows)
```

```python
print(len(myrows))
```

```python
myMovies_List = []

for row in myrows:
    dict={}
    dict['Rank']=row.find_all('td')[0].text.strip()
    dict['Title']=row.find_all('th')[0].text.strip()
    dict['Worldwide gross']=row.find_all('td')[1].text.strip()
    dict['Primary language(s)']=row.find_all('td')[2].text.strip()
    dict['Year']=row.find_all('td')[3].text.strip()
    myMovies_List.append(dict)
```

```python
print(myMovies_List)
```

```python
Dataset = pd.DataFrame(myMovies_List)
```
