```python
df = pd.read_excel(r'E:\Data Analyst Roadmap\Google Data Analytics Certificate Course\Python\Alex the Analyst Playlist of Python\Pandas\Movie-Data.xlsx')
```

```python
df
```

```python
Count_Movies_By_Genre = df.groupby('Genre').agg({'Movie Title':'count'})
```

```python
Count_Movies_By_Genre.plot(kind='line',figsize=(10,5),xlabel='Genre',ylabel='No of Movies',title='Movies Distribution By Genre',grid=True)
```

```python
Count_Movies_By_Genre.plot(kind='pie',y='Movie Title',figsize=(15,8),title='Movies Distribution By Genre')
```

```python
df
```

```python
top_20_Movies_By_Revenue = df.groupby('Movie Title').agg({'Revenue':'sum'}).sort_values(by='Revenue',ascending=False)
```

```python
visual_top_20_movies_by_Revenue = top_20_Movies_By_Revenue.head(20)
```

```python
visual_top_20_movies_by_Revenue
```

```python
visual = visual_top_20_movies_by_Revenue.plot(kind='barh',xlabel='Revenue in Crores',title='top 20 Movies By Revenue Lowest to Highest',figsize=(8,8))
```

```python
print(plt.style.available)
plt.style.use('ggplot')
```
