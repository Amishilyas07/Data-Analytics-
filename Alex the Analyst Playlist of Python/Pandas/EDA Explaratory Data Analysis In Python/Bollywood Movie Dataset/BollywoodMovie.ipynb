```python
melted_Genres = df.melt(id_vars=['imdbId', 'hitFlop'], value_vars=['Genre1', 'Genre2', 'Genre3'], value_name='Genre')

# Strip whitespace from the Genre values
melted_Genres['Genre'] = melted_Genres['Genre'].str.strip()

# Filter for successful genres (e.g., Blockbuster)
successful_genres = melted_Genres[melted_Genres['hitFlop'].str.contains('Block')]

# Group by Genre and count occurrences
genre_counts = successful_genres.groupby('Genre').agg({'imdbId': 'count'}).reset_index()

# Rename columns for clarity
genre_counts.columns = ['Genre', 'Count']

# Sort the genres by count in descending order
genre_counts = genre_counts.sort_values(by='Count', ascending=False)

# Display the resulting DataFrame
genre_counts
```
