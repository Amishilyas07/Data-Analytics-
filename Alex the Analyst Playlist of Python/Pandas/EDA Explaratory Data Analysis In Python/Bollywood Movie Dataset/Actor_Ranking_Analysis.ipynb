```python

top_10_actors = df.sort_values(by='movieCount', ascending=False).head(10)
# Extract actor names and movie counts
most_Movies_Done = top_10_actors['actorName'].tolist()
actor_done_Most_Movies = top_10_actors['movieCount'].tolist()

# Plotting a pie chart
plt.figure(figsize=(5, 8))
plt.pie(actor_done_Most_Movies, labels=most_Movies_Done, autopct='%1.1f%%')
plt.title('Top 10 Actors by Movies Count 2001 - 2014')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()

```

```python
# Actors Popularity vs Rating Correlation
plt.figure(figsize=(10,6))
plt.scatter(df['googleHits'],df['ratingSum'],alpha=0.5)
plt.title('Actors Popularity Vs Rating Correlation')
plt.xlabel('Popularity')
plt.ylabel('Rating Sum')
plt.show()
```
