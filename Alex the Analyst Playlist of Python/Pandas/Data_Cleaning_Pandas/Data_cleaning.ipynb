```python
# Removing Duplicates 
df = df.drop_duplicates()
```

```python
df
```

```python
# droppig Columns 
df = df.drop(columns='Not_Useful_Column')
```

```python
df
```

```python
# lstrip
df['Last_Name'] = df['Last_Name'].str.lstrip('/')
df['Last_Name'] = df['Last_Name'].str.lstrip('...')
```

```python
df
```

```python
# rstrip
df['Last_Name']=df['Last_Name'].str.rstrip('_')
```

```python
df
```

```python
# Other Method Of Removing things from column
df['Last_Name'] = df['Last_Name'].str.strip('123._/')
```

```python
df
```

```python
# Standarizing Phone Numbers 
# this will Replace Other Characters that are mentioned except them with empty String
df['Phone_Number'] = df["Phone_Number"].replace('[^a-zA-Z0-9]', '', regex=True)
```

```python
df
```

```python
# Now Apply Dashes in Number with Lambda 
# First Convert Numbers to String then apply function
df['Phone_Number'] = df['Phone_Number'].astype('str')
df['Phone_Number']=df['Phone_Number'].apply(lambda x : x[0:3] + '-' + x[3:6] + '-' + x[6:10])
```

```python
df
```

```python
# Now Replace nan--  and Na-- with Empty String
df['Phone_Number'] = df['Phone_Number'].replace('nan--','')
df['Phone_Number'] = df['Phone_Number'].replace('Na--','')
```

```python
df
```

```python
# Splitting Columns Based on Comma
# Adding New Columns Strees Address , State , Zipcode that will recieve values according to index
df[['Street Address','State','Zipcode']] = df['Address'].str.split(',',n=2,expand=True)
```

```python
df
```

```python
df
```

```python
# Replcing the Y to Yes and N to No 
df['Paying Customer']=df['Paying Customer'].replace('Y','Yes',regex=False)
df['Paying Customer']=df['Paying Customer'].replace('N','No',regex=False)
df['Do_Not_Contact']=df['Do_Not_Contact'].replace('Y','Yes',regex=False)
```

```python
df
```

```python
# Drop Null Values 
df.fillna('')
```

```python
# Remove Rows or Records Where Do Not Contact is Yes
for x in df.index:
    if df.loc[x,'Do_Not_Contact']=='Yes':
        df.drop(x,inplace=True)
df        
```
