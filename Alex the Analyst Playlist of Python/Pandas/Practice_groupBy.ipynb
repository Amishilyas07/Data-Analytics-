```python
Top_10_Movies_By_Budget = df.groupby('Movie Title').sum('Budget')
```

```python
Top_10_Movies_By_Budget.sort_values(by='Budget',ascending=False)
```

```python
Top_10_Movies_By_Budget.sort_values(by='Budget',ascending=False).head(10)
```

```python
top_10_Movies_By_Revenue = df.groupby('Movie Title').agg({'Revenue':'sum'})
```

```python
top_10_Movies_By_Revenue.sort_values('Revenue',ascending=False).head(10)
```

```python
top_10_Movies_With_Lowest_Budget_Highest_Revenue = df.groupby(['Movie Title','Genre']).agg({'Budget':'sum','Revenue':'sum'})
```

```python
top_10_Movies_With_Lowest_Budget_Highest_Revenue.sort_values(by=['Budget','Revenue'],ascending=[True,False]).head(10)
```
